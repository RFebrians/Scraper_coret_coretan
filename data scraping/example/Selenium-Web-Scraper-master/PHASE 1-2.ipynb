```python
# importing necessary packages
from selenium import webdriver
from webdriver_manager.chrome import ChromeDriverManager

# for holding the resultant list
element_list = []

for page in range(1, 3, 1):
	
	page_url = "https://www.siloamhospitals.com/cari-dokter/rumah-sakit/siloam-hospitals-sentosa"
	driver = webdriver.Chrome(ChromeDriverManager().install())
	driver.get(page_url)
	title = driver.find_elements_by_xpath("(//div[@class='r-result'])[1]")
	price = driver.find_elements_by_xpath("//div[@class='r-result']//div[2]//div[2]//table[1]")
    

	for i in range(len(title)):
		element_list.append([title[i].text, price[i].text])

print(element_list)

#closing the driver
driver.close()
```
