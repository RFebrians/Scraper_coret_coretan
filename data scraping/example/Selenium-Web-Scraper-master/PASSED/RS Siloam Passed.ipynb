```python
# importing necessary packages
from selenium import webdriver
from webdriver_manager.chrome import ChromeDriverManager

# for holding the resultant list
element_list = []

for page in range(1, 3, 1):
	
	page_url = "https://www.halodoc.com/cari-dokter/spesialis/obstetrics-and-gynecologist"
	driver = webdriver.Chrome(ChromeDriverManager().install())
	driver.get(page_url)
	title = driver.find_elements_by_xpath("//div[@class='col-12 col-md-8 col-xl-9 hospital-list--result ng-star-inserted']")   

	for i in range(len(title)):
		element_list.append([title[i].text])
        

print(element_list)

#closing the driver
driver.close()
```
