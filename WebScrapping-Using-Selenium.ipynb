```python
import pandas as pd
import random
from time import sleep
from selenium import webdriver

from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.by import By
from selenium.common.exceptions import NoSuchElementException
```

## So, let's procced to code making it more interactive

```python
# For using Selenium we must call the driver, also is recommended to accept the cookings
driver = webdriver.Chrome('/home/zegveld/Downloads/chromedriver')
driver.get('https://www.mercadolibre.com.co/')
sleep(random.uniform(3.0, 5.0))
WebDriverWait(driver,100).until(EC.element_to_be_clickable((By.CSS_SELECTOR,"button#cookieDisclaimerButton"))).click()

# Because the link is in a box and does not have a exactly Xpath we must make a loop to find the text to which we are going to go.
a_tags = driver.find_elements_by_xpath('//a[@class="category"]')


for a in a_tags:
    p_tag = a.find_element_by_tag_name('p')
    if p_tag.text == 'Computación':
        result = a
        break

# Also it's really important to setup the random times to give the coding enough time get the data
sleep(random.uniform(3.0, 5.0))
result.click()


# Now we could proceed to link in the Computers Category
driver.find_element_by_xpath('//h3[@class="title"]').click()

# To continue with the data collection, lists are created

main_list= []
boton = driver.find_element_by_xpath('//span[text()="Siguiente"]')
sleep(random.uniform(3.0, 5.0))
# In this step it is required to create a loop to access the information depending on the boxes content an ther class link.

for i in range(3):
    main_box = driver.find_elements_by_xpath('//div[@class="ui-search-result__content-wrapper"]')
    
# In ths loop the price and the product name is extracted from the main box information of each one.
    
    for pc in main_box:
        precio = pc.find_element_by_xpath(
        './/span[@class="price-tag-fraction"]').get_attribute("innerHTML")
        descripcion = pc.find_element_by_xpath(
        './/h2[@class="ui-search-item__title"]').get_attribute("innerHTML")

#  Some of the products does not pay VAT (Value Added Tax), so we extract which of them do not pay it.      
        try:
            vat_tag = pc.find_element_by_xpath(
            './/span[@class="ui-search-styled-label ui-search-item__highlight-label__text"]').get_attribute("innerHTML")
        
            if vat_tag == 'MÁS VENDIDO':
                  vat_tag = None
        except NoSuchElementException:
                vat_tag = None
            
        
        main_list.append({'Product Description': descripcion, 
                          'Price' : precio,
                          'VAT exclusion': vat_tag})        
        
    boton.click()
    sleep(random.uniform(3.0, 5.0))
    boton = driver.find_element_by_xpath('//span[text()="Siguiente"]')

# Finally the information is stored
driver.quit()
```

```python
# For using Selenium we must call the driver, also is recommend to accept the cookings

driver = webdriver.Chrome('/home/zegveld/Downloads/chromedriver')
driver.get('https://www.mercadolibre.com.co/')
sleep(random.uniform(3.0, 5.0))
WebDriverWait(driver,20).until(EC.element_to_be_clickable((By.CSS_SELECTOR,"button#cookieDisclaimerButton"))).click()


# Because the link is in a box and does not have a exactly Xpath we must make a loop to find the text to which we are going to go.

a_tags = driver.find_elements_by_xpath('//a[@class="category"]')


for b in a_tags:
    b_tag = b.find_element_by_tag_name('p')
    if b_tag.text == 'Celulares y Teléfonos':
        result2 = b
        break

# Also it's really important to settup the random times to give the coding enough time get the data

result2.click()
sleep(random.uniform(4.0, 5.0))
# Now we could proceed to link in the Phones Category
driver.find_element_by_xpath('//a[@class="dynamic__carousel-link"]').click()
sleep(random.uniform(3.0, 5.0))



boton = driver.find_element_by_xpath('//span[text()="Siguiente"]')

#In this step it is required to create a loop to access the information of each click
for i in range(2):
    
    info = driver.find_elements_by_xpath('//div[@class="ui-search-result__content-wrapper"]')        

    for phone in info:
        precio = phone.find_element_by_xpath(
        './/span[@class="price-tag-fraction"]').text
        descripcion = phone.find_element_by_xpath(
        './/h2[@class="ui-search-item__title"]').text
        
        try:
            vat_tag = phone.find_element_by_xpath(
            './/span[@class="ui-search-styled-label ui-search-item__highlight-label__text"]').text
        
            if vat_tag == 'MÁS VENDIDO':
                  vat_tag = None
        except NoSuchElementException:
                vat_tag = None
            
        
        main_list.append({'Product Description': descripcion, 
                          'Price' : precio,
                          'VAT exclusion': vat_tag})        
         
    boton.click()
    sleep(random.uniform(4.0, 6.0))
    boton = driver.find_element_by_xpath('//span[text()="Siguiente"]')
    
driver.quit()
```
